```python
import cv2
import numpy as np
import pandas as pd
from numpy import dot
from numpy.linalg import norm
```

```python
cap = cv2.VideoCapture("2.avi")

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
#fps = int(cap.get(cv2.CAP_PROP_FPS))
fourcc = cv2.VideoWriter_fourcc(*'DIVX')

out = cv2.VideoWriter("2output.avi", fourcc, 30, (frame_width, frame_height))
```

```python
pre = []
count = 0
dd = []
```

```python
while cap.isOpened():
    ret, frame = cap.read()

    if ret == True:

        frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        hist = cv2.calcHist(frame_rgb, [0, 1, 2], None, [8, 8, 8], [0, 256, 0, 256, 0, 256])    
        hist = cv2.normalize(hist, hist).flatten()
        if count != 0:
            d = cv2.compareHist(pre, hist, cv2.HISTCMP_BHATTACHARYYA)
            if d >= 0.16:
                out.write(frame)
            dd.append(d)
        pre = hist
        count = count + 1
        cv2.imshow("Image", frame_rgb)
        if cv2.waitKey(1) & 0xFF == ord('s'):
            break
cap.release()
out.release()
cv2.destroyAllWindows()
```
